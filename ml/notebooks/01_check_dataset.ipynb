```python
# check missing value 
df.isnull().sum()
for i in df:
    if df[i].isnull().sum() > 0:
        print(f"{i} has {df[i].isnull().sum()} missing values")
        if df[i].isnull().sum() > 0.6 * len(df):
            print(f"{i} has more than 60% missing values, consider dropping this column")
            # drop the column if it has more than 60% missing values
            
            
        
```

```python
for i in df.columns.tolist():
    if df[i].isnull().sum() > 0.6 * len(df):
        df.drop(columns=[i], inplace=True)
        print(f"{i} has been dropped")
```

```python
#check catogorical columns
catogorical_col=df.select_dtypes(include="object").columns
catogorical_col
for i in catogorical_col:
    print("missing value:",df[i].isna().sum())
    # fillmissing value using mode 
    mode=df[i].mode()[0]
    df[i]=df[i].fillna(mode)
    print(f"missing value will fill{df[i]}")
```

```python
#check numerical columns
numerical_col=df.select_dtypes(include='int64').columns
for i in numerical_col:
    print("missing value:",df[i].isna().sum())
```

```python
# check outliear 
# Select numerical columns
numerical_columns = df.select_dtypes(include="number").columns

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\nColumn: {col}")
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Number of outliers:", len(outliers))
```

```python
# remove outliear 
for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Clip outliers
    df[col] = df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outliers clipped successfully!")
```

```python
import os

# Move from notebooks/ to the project root, then into data/raw
folder_path = os.path.join("..", "data", "raw")

os.makedirs(folder_path, exist_ok=True)

df.to_csv(
    os.path.join(folder_path, "cleane_hotel.csv"),
    index=False
)

print("Dataset saved successfully!")
```
